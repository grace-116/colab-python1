```python
# 1. 람다 표현식으로 함수 만들고 호출하기
calculate_power = lambda x, y: x ** y
print("람다 할당 호출:", calculate_power(2, 3))

# 2. 람다 표현식 자체를 즉시 호출하기 (IIFE 방식)
result = (lambda x, y: x + y)(10, 20)
print("람다 즉시 호출:", result)

# 3. 람다 표현식 안에서 변수 생성 불가 (아래 주석 해제 시 SyntaxError 발생)
# (lambda x: y = 10; x + y)(1)

# 단, 람다 표현식 외부의 변수는 자유롭게 참조 가능합니다.
offset = 5
print("외부 변수 참조:", (lambda x: x + offset)(10))
```

# 3. map, filter, reduce 함수와 람다 표현식의 결합

데이터 전처리나 배열 연산 시 반복문(`for`, `while`)을 쓰지 않고도 람다 표현식과 내장 함수를 결합하여 처리 속도와 가독성을 높일 수 있습니다.

* **`map(함수, 반복가능객체)`**: 리스트나 튜플의 모든 요소에 지정된 함수를 일괄 적용합니다.
* **`filter(함수, 반복가능객체)`**: 특정 조건을 만족하는(결과가 참인) 요소만 걸러냅니다.

```python
from functools import reduce

# 가상의 센서 측정 데이터
sensor_values = [1.2, -0.5, 2.3, -1.1, 3.0]

# 1. map 활용: 모든 센서 값에 증폭률(10) 곱하기
scaled_values = list(map(lambda x: x * 10, sensor_values))
print("map 처리 결과 (스케일링):", scaled_values)

# 2. filter 활용: 노이즈(- 값)를 제외하고 0보다 큰 정상 데이터만 추출하기
valid_values = list(filter(lambda x: x>1, sensor_values))
print("filter 처리 결과 1 이상(정상값 추출):", valid_values)
```

```python
raw_data = [15, -2, 23, -5, 30]

# 1. 조건부 표현식 활용: 0 미만인 값은 0으로 일괄 보정, 0 이상은 그대로 유지
# 형식: lambda 매개변수: 식1 if 조건식 else 식2
corrected_data = list(map(lambda x: x if x >= 0 else 0, raw_data))
print("조건부 람다 처리 결과:", corrected_data)

# 2. map에 여러 객체 넣기: 두 개의 센서 배열 값 합산하기
sensor_a = [1, 2, 3, 4, 5]
sensor_b = [10, 20, 30, 40, 50]

# sensor_a의 요소는 x로, sensor_b의 요소는 y로 동시에 전달됩니다.
combined_data = list(map(lambda x, y: x + y, sensor_a, sensor_b))
print("다중 객체 map 처리 결과:", combined_data)
```

```python
# 1. 전역 변수 (Global Variable)
# 프로그램 어디서나 접근할 수 있는 변수입니다.
app_name = "MyPythonApp"

def greet_user():
    # 2. 지역 변수 (Local Variable)
    # 함수 내부에서만 존재하며, 함수가 종료되면 메모리에서 사라집니다.
    user_name = "Alice"

    # 함수 내부에서는 전역 변수와 지역 변수 모두 접근 가능합니다.
    print(f"[{app_name}] 안녕하세요, {user_name}님!")

# 함수 호출
greet_user()

# 3. 변수 접근성 테스트
print("--- 함수 외부에서 접근 ---")
print(f"앱 이름: {app_name}")  # 전역 변수이므로 성공

# 아래 코드는 NameError를 발생시킵니다.
# print(f"사용자 이름: {user_name}")
```

```python
score = 0

def update_score_fail():
    # 전역 변수 score와 이름만 같은 '지역 변수'가 새로 만들어집니다.
    # 함수 밖의 score에는 아무런 영향을 주지 못합니다.
    score = 100
    print(f"함수 내부(실패): {score}")

def update_score_success():
    # 함수 밖에 있는 전역 변수 score를 사용하겠다고 명시적으로 선언합니다.
    global score
    score += 10
    print(f"함수 내부(성공): {score}")

# 실행 결과 확인
print(f"현재 점수: {score}")
update_score_fail()
print(f"실패 함수 실행 후 점수: {score}") # 여전히 0

update_score_success()
update_score_success()
print(f"성공 함수 2회 실행 후 점수: {score}") # 20으로 업데이트 완료
```

#클래스의 핵심 개념

| 용어 | 설명 | 비유 |
| :--- | :--- | :--- |
| **클래스** | 객체를 만들기 위한 설계도입니다. | 붕어빵 틀 |
| **인스턴스** | 설계도로부터 실제로 만들어진 결과물입니다. | 구워져 나온 붕어빵 |
| **속성** | 객체가 가지는 데이터입니다. | 팥, 슈크림 등 내용물 |
| **메서드** | 객체가 수행하는 함수입니다. | 붕어빵을 굽는 동작 |

---

###클래스 사용의 장점

*   **코드 재사용**: 한 번 정의한 클래스로 무수히 많은 객체를 생성할 수 있습니다.
*   **유지보수 용이**: 데이터와 기능이 하나로 묶여 있어 코드 수정 시 영향 범위를 파악하기 쉽습니다.
*   **구조적 설계**: 프로그램의 구성 요소를 독립적인 단위로 나누어 설계할 수 있습니다.

> **Note**
> 클래스는 단순히 데이터를 담는 그릇을 넘어, 데이터와 로직을 긴밀하게 결합하여 코드의 복잡성을 획기적으로 줄여줍니다.

```python
# 1. 클래스와 메서드 정의하기
class VibrationSensor:
    # 메서드를 정의할 때는 첫 번째 자리에 반드시 self를 적습니다.
    def measure(self):
        print("진동 데이터를 측정합니다.")

    def calibrate(self):
        print("센서 영점을 조절합니다.")

# 2. 클래스를 통해 인스턴스(객체) 생성하기
sensor_a = VibrationSensor()

# 3. 객체의 메서드 호출하기
# 호출할 때는 self 자리를 파이썬이 알아서 채워주므로 빈칸으로 둡니다.
sensor_a.measure()
sensor_a.calibrate()
```

```python
# 3. 딕셔너리(dict) 클래스의 인스턴스 생성 및 메서드 활용
# 중괄호를 사용하는 것은 dict 클래스의 인스턴스를 만드는 단축 문법입니다.
sensor_config = {"channel": 1, "sampling_rate": 1000}

# dict 클래스에 내장된 keys 메서드와 get 메서드를 호출합니다.
print(sensor_config.keys())
print(sensor_config.get("sampling_rate"))
print(type(sensor_config))
```
