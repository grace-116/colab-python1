```python
# 파일에 문자열 쓰기
# w는 기존 파일 내용을 완전히 삭제하고 새로 작성
# 해당 파일이 없으면 새로 생성
file = open('oha.txt','w')
file.write('Hello guys.')
file.close()

# 파일에서 문자열 읽기
file = open('oha.txt', 'r')
text = file.read()
print(text)
file.close()

# with 구문 - 파일 열고 자동으로 닫기
with open('oha.txt', 'r') as file:
    text = file.read()
    print(text)

# 여러줄의 문자열 파일에 쓰기
with open('data_log.txt', 'w') as file:
    for i in range(5):
        file.write(f'Data point {i}\n')
```

```python
# writelines - 리스트의 문자열을 파일에 쓰기
lines = ['전력\n', '현재\n', '저항\n']
with open('params.txt', 'w') as file:
    file.writelines(lines)

# readlines - 파일의 내용을 한 줄씩 리스트로 가져오기
with open('params.txt', 'r') as file:
    lines_list = file.readlines()
    print(lines_list)

# strip - 파일의 내용을 줄 단위로 순회하기
with open('params.txt', 'r') as file:
    for line in file:
        print(line.strip())
```

```python
import pickle

# 저장할 딕셔너리 데이터
config = {'반복': 1000, '숫자': 1e-5}

# 객체를 바이너리 파일로 저장
with open('config.p', 'wb') as file:
    pickle.dump(config, file)

# 바이너리 파일에서 객체 복원
with open('config.p', 'rb') as file:
    config_read = pickle.load(file)
    print(config_read)
```

# 파일을 연 뒤에 반드시 닫아야 하는 이유


1. 자원 누수 방지
2. 데이터 무결성 및 버퍼 플러시
3. 파일 잠금 해제

```python
# 실험 결과인 [10.5, 20.3, 15.8] 리스트를 raw_data.txt 파일에 한 줄에 하나씩 기록하시오.
# 그 다음 해당 파일을 다시 읽어 들여 모든 숫자의 합계를 구하고 출력하시오.
'''
조건 1: 파일 열기 및 닫기는 with 구문을 사용하시오.
조건 2: 문자열을 숫자로 변환하는 과정이 필요니다.
'''

measurements = [10.5, 20.3, 15.8]

# 1. measurements 리스트의 값을 한 줄씩 raw_data.txt에 쓰기
with open('raw_data.txt', 'w') as file:
    for i in measurements:
        file.write(str(i) + '\n')

# 2. raw_data.txt를 읽어들여 모든 값의 합계 계산 및 출력
total_sum = 0

with open('raw_data.txt', 'r') as file:
    for line in file:
        total_sum += float(line.strip())

print("합계:", total_sum)
```

| 함수 및 메서드 | 기능 설명 |
|---|---|
| open(file, mode) | 지정한 파일 이름과 동작 모드(r, w, rb, wb)에 따라 파일 객체를 생성하여 반환합니다. |

| read() | 파일 내부의 전체 텍스트 데이터를 단일 문자열 데이터형으로 한 번에 읽어옵니다. |

| pickle.dump(obj, file) | 파이썬 고유의 객체 구조(리스트, 딕셔너리 등)를 바이너리 데이터 형태로 파일에 저장합니다. |
| pickle.load(file) | 바이너리 파일로부터 저장된 데이터를 읽어 원래의 파이썬 객체 구조로 복원합니다. |

```python
def Adata(data):
    # 데이터가 비어있을 경우 0을 3개 반환
    if len(data) == 0:
        return 0, 0, 0

    max_ = max(data)
    min_ = min(data)
    avg_ = sum(data) / len(data)

    # 3개의 값을 쉼표로 구분하여 반환 (자동으로 튜플로 패킹됨)
    return max_, min_, avg_

# 테스트용 센서 데이터 리스트
sensor = [22.5, 23.1, 21.8, 24.2, 22.9]

# 함수 호출 및 3개의 변수로 직접 언패킹하여 받기
highest, lowest, average = Adata(sensor)

print(f"최고 온도: {highest}")
print(f"최저 온도: {lowest}")
print(f"평균 온도: {average:.1f}")
```

```python
# 두 정수 a와 b를 매개변수로 받아서 덧셈, 뺄셈, 곱셈, 나눗셈(몫)의 결과를 한 번에 반환하는 함수 calc를 작성하시오
#  b가 0이면 연산을 수행하지 않고 즉시 종료되도록 구현하시오.

def calc(a, b):
  if b == 0:
    return

  res_add = a + b
  res_sub = a - b
  res_mul = a * b
  res_div = a // b

  return res_add, res_sub, res_mul, res_div

res_add, res_sub, res_mul, res_div = calc(10, 2)
print(res_add, res_sub, res_mul, res_div)
```

```python
# 1. 위치 인수와 리스트 언패킹
def numbers(a, b, c):
    print('a:', a)
    print('b:', b)
    print('c:', c)
    print('-' * 15)

# 기본적인 위치 인수 전달
numbers(10, 20, 30)

# 리스트 언패킹을 사용한 전달
num_list = [100, 200, 300]
numbers(*num_list)

# 튜플도 동일하게 언패킹 가능
num_tuple = (7, 8, 9)
numbers(*num_tuple)

'''
매개변수 개수가 다르면 에러가 일어남
wrong_list = [1, 2]
numbers(*wrong_list)
'''

# 가변 인수 함수 정의
def calculate_sum(*args):
    print('입력된 가변 인수 튜플:', args)

    total = 0
    for num in args:
        total += num
    return total

print('합계:', calculate_sum(3,4,5))
print('합계:', calculate_sum(10, 20, 30, 40, 50))
print('-' * 15)

# 고정 매개변수와 가변 인수의 혼합 사용
# 반드시 고정 매개변수를 먼저 선언하고, 가변 인수를 맨 뒤에 두어야 합니다.
def print_student_scores(name, *scores):
    print('학생 이름:', name)
    print('과목 점수들:', scores)

print_student_scores('홍길동', 85, 90, 95)
```

```python
# 2. 키워드 인수와 딕셔너리 언패킹
def personal_info(name, age, address):
    print('이름:', name)
    print('나이:', age)
    print('주소:', address)
    print('-' * 15)

# 키워드 인수를 사용하면 순서를 바꿔도 상관없음 (=)
personal_info(age=30, address='서울', name='김철수')

# 딕셔너리 언패킹 (**)
user_dict = {'name': '이영희', 'age': 25, 'address': '부산'}
personal_info(**user_dict)

# 가변 키워드 인수 함수 정의
def create_profile(**kwargs):
    print('전달받은 딕셔너리:', kwargs)

    for x, y in kwargs.items():
        print(f'{x}: {y}')
    print('-' * 15)

# 어떤 키워드를 넣어도 모두 딕셔너리로 묶여서 처리됩니다.
create_profile(name='박지민', job='데이터 엔지니어')
create_profile(name='최수연', age=28, city='인천', hobby='수영')
```

```python
# 모든 인수를 종합한 궁극의 함수 형태 예시
def master_func(a, b=10, *args, **kwargs):
    print('a (일반):', a)
    print('b (초깃값):', b)
    print('args (가변 위치):', args)
    print('kwargs (가변 키워드):', kwargs)

master_func(1, 2, 3, 4, 5,6, x=100, y=200)
```

1. **위치 인수 구역 (`a, b=10`)**: 여기서는 자리가 중요하므로 반드시 초깃값이 없는 것(`a`)이 먼저, 있는 것(`b=10`)이 나중에 와야 합니다.
2. **키워드 전용 인수 구역 (`c, d=20`)**: `*args` 뒤에 있는 매개변수들은 어차피 무조건 **이름표(키워드)를 달고 값을 전달받아야 합니다.** 이름표가 있기 때문에 파이썬 인터프리터가 순서나 자리를 헷갈릴 일이 없습니다.

따라서 `*args` 뒤에서는 초깃값이 있든 없든 섞어서 정의해도 아무런 문제가 발생하지 않습니다. 단지 `c`는 **"반드시 이름을 적어서 값을 넣어줘야 하는 필수 키워드 인수"**가 될 뿐입니다.

```python
def check_difference(a, b=10, *args, c, d=20):
    print(f"b의 값: {b}")
    print(f"d의 값: {d}")
    print(f"args에 흡수된 값들: {args}")
    print("-" * 20)

check_difference(1, 99, 2, 3, 4, c=100)

check_difference(1, 99, 2, 3, 4, c=100, d=77)

check_difference(1, c=100, b=99, d=77)

#check_difference(1, 99, 2, 3, 4) -> c의 값이 명시되지 않음
```

# 정의(Definition)와 호출(Call)의 명확한 구분 (혼동 주의)


**1. 함수를 정의할 때 (매개변수, Parameter)**
정의할 때는 변수의 이름과 초깃값 여부만 결정합니다.
* `a`: 초깃값이 없는 매개변수 (무조건 값을 받아야 함)
* `b=10`: 초깃값이 있는 매개변수 (값을 안 받으면 10을 씀)
* 이때는 `a`나 `b`를 위치 인수나 키워드 인수라고 부르지 않습니다. 둘 다 호출 방식에 따라 위치로도, 키워드로도 값을 받을 수 있기 때문입니다.

**2. 함수를 호출할 때 (인수, Argument)**
정의된 함수에 실제로 값을 던져넣을 때 비로소 '위치'와 '키워드'의 개념이 생깁니다.
* `func(1, 2)`: 순서에 맞춰서 던졌으니 1과 2는 **위치 인수**입니다. (1은 a로, 2는 b로 들어감)
* `func(a=1, b=2)`: 이름표를 붙여서 던졌으니 a=1과 b=2는 **키워드 인수**입니다.

```python
def perfect_order_func(a, b=10, *args, c, d=20, **kwargs):
    print(a)       # 1. 일반 위치 인수
    print(b)       # 2. 기본값이 지정된 위치 인수
    print(args)    # 3. 가변 위치 인수
    print(c)       # 4. 키워드 전용 인수 (기본값 없음)
    print(d)       # 4. 키워드 전용 인수 (기본값 있음)
    print(kwargs)  # 5. 가변 키워드 인수

perfect_order_func(1, 2, 3, 4, 5, c=100, x=200, y=300)
```

```python
# 여러 센서 장비의 이름과 상태 코드를 가변 키워드 인수로 입력받아
# 에러가 발생한 장비의 이름만 리스트로 반환하는 find_error_equipment 함수를 작성하시오.
'''
조건 1: 인수는 가변 키워드 인수 kwargs 하나만 사용합니다.
조건 2: 상태 코드가 문자열 Error 인 장비의 이름(키)만 추출하여 리스트에 담습니다.
조건 3: 에러가 발생한 장비가 없다면 빈 리스트를 반환합니다.
'''

def find_error_equipment(**kwargs):
    error_list = []

    for x,y in kwargs.items():
        if y == 'Error':
            error_list.append(x)
    return error_list

result1 = find_error_equipment(sensor_a='Normal', sensor_b='Error', sensor_c='Normal', motor_x='Error')
print('에러 발생 장비:', result1)

result2 = find_error_equipment(valve_1='Normal', valve_2='Normal')
print('에러 발생 장비:', result2)
```

```python
# 종료 조건을 추가한 재귀호출
def hello_count(count):
    if count == 0:    # 종료 조건: count가 0이면 더 이상 호출하지 않음
        return

    print('Hello, world!', count)

    count -= 1        # 재귀 호출을 할 때마다 상태를 변경
    hello_count(count) # 자기 자신을 호출

hello_count(5)
```

```python
def factorial(n):
    # 1. 종료 조건: n이 1일 때 재귀를 멈추고 1을 반환
    if n == 1:
        return 1

    # 2. 재귀 단계: n과 factorial(n - 1)의 곱을 반환
    return n * factorial(n - 1)

print('5 팩토리얼 결과:', factorial(5))
```

```python
def fibonacci(n):
    # 종료 조건: 1번째 또는 2번째 피보나치 수는 1
    if n == 1 or n == 2:
        return 1

    # 앞의 두 피보나치 수를 더하는 재귀 호출
    return fibonacci(n - 1) + fibonacci(n - 2)

# 6번째 피보나치 수 계산 (1, 1, 2, 3, 5, 8) -> 8 출력
print('6번째 피보나치 수:', fibonacci(6))
```
